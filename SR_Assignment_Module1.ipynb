**Write a python program to reverse words of sentence in the same order

```python
def reverse_words(sentence: str):
    # Split sentence into words
    words = sentence.split()
    print(words)
    # Reverse each word
    reversed_words = [word[::-1] for word in words]
    # Form the sentence from individual reversed words
    print(reversed_words)
    return " ".join(reversed_words)


input= "Question one is successfully solved"
result = reverse_words(input)

print("Input: ", input)
print("Output:", result)
```

**Write a python code to generate the powerset of a given set

```python
from itertools import combinations

def powerset(s: list):
    result = []
    # Loop over all possible subset lengths from 0 to len(s)
    for r in range(len(s) + 1):
        # Generate combinations of length r and convert tuples to lists
        for combo in combinations(s, r):
            result.append(list(combo))
    return result


# Example Usage
input_set = [1, 2, 3]
output = powerset(input_set)

print(f"Number of Powerset for the set {input_set} is 2^{len(input_set)} which equals {len(output)} \n")
print("Powersets : \n")
for subset in output:
    print(subset)
```

Write a python code to pickup a list from a string dynamically

```python
import re

def extract_list(text: str):
    # Match content inside the first pair of square brackets []
    match = re.search(r"\[(.*?)\]", text)
    #print(match)

    if not match:
        return []

    # Get the raw string inside brackets
    raw_content = match.group(1)
    #print(raw_content)

    

    # Split by comma and strip extra whitespace/quotes from items
    items = [item.strip().strip("'\"") for item in raw_content.split(",")]
    return items


# Example Usage
sentence = "The colours of jackets in the shop is ['Blue', 'Indigo', 'Green']"
extracted_list = extract_list(sentence)

print(f"Input Sentence: {sentence}")
print(f"Extracted List: {extracted_list}")
print(f"List Count: {len(extracted_list)}")
```

```python
def chunk_text(text: str, chunk_size: int, overlap_size: int) -> list[str]:
    """Divides text into chunks of specified size with overlap.

    Args:
        text (str): The input string to split.
        chunk_size (int): The target length of each chunk.
        overlap_size (int): The number of characters to overlap between
          consecutive chunks.

    Returns:
        list[str]: A list of text chunks.
    """
    # Validation checks
    if chunk_size <= 0:
        raise ValueError("chunk_size must be greater than 0.")
    if overlap_size < 0:
        raise ValueError("overlap_size cannot be negative.")
    if overlap_size >= chunk_size:
        raise ValueError("overlap_size must be strictly less than chunk_size.")

    chunks = []
    step = chunk_size - overlap_size
    start = 0
    text_len = len(text)

    # Loop through the text creating sliding windows
    while start < text_len:
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)

        # Move the sliding window forward by the step size
        start += step

    return chunks


# Example Usage
sample_text = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
c_size = 10
o_size = 3

result = chunk_text(sample_text, chunk_size=c_size, overlap_size=o_size)

print(f"Original Text: {sample_text}")
print(f"Chunk Size: {c_size}, Overlap Size: {o_size}\n")
for i, chunk in enumerate(result, 1):
    print(f"Chunk {i}: '{chunk}'")
```
